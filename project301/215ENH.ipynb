**How Fingerprint Recognition Works:** Just as cellular automata rely on local patterns to determine global behavior, fingerprint recognition depends on analyzing local ridge patterns to identify individuals. The unique patterns in a fingerprint—specifically ridge endings and bifurcations (collectively called minutiae)—act like the cells in a CA grid. Where a CA uses neighboring cell states to compute the next generation, SIFT uses neighboring pixels to detect and describe these distinctive features. The system then compares these feature sets across fingerprints, similar to how correlation slides a kernel across an array to find matches.

We then need to extract the zip file so that the .tif image files become accessible and the image file paths can be processed. Since there are many different folders in this archive, we focus on images from one folder 'Db1_a', which contains a subset of images for the year 2000.

```python
#Written with assistance from ChatGPT
def get_finger_print(zip_file, target_folder='Db1_a'):
    """
    Extracts the zip file and returns the path of the fingerprint images.
    """
    extract_dir = 'fingerprint_data'

    if os.path.exists(extract_dir):
        shutil.rmtree(extract_dir)

    with zipfile.ZipFile(zip_file, 'r') as zip_ref:
        zip_ref.extractall(extract_dir)

    finger_prints = []
    for root, _, files in os.walk(extract_dir):
        if target_folder in root:
            for file in files:
                if file.lower().endswith('.tif'):
                    finger_prints.append(os.path.join(root, file))

    return finger_prints
```

Now that we have the fingerprint images we want to use, they must be processed to ensure that they are in a format suitable for comparison in later steps, and to ensure reliable results.

```python
#Written with assistance from ChatGPT
def load_fingerprint(path):
    """
    Loads and reads a fingerprint image from inputed path and verifies it is in a suitable format by
    converting it to greyscale and normalizing the pixel values (8-bit). The processed image is then
    returned.
    """
    img = cv2.imread(path, cv2.IMREAD_UNCHANGED)

    if img is None:
        raise ValueError(f"Could not read image: {path}")

    if len(img.shape) == 3:
        img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    if img.dtype != np.uint8:
        img = cv2.normalize(img, None, 0, 255, cv2.NORM_MINMAX)
        img = img.astype(np.uint8)

    return img
```

**Why SIFT?** The Scale-Invariant Feature Transform (SIFT) works by identifying key points in an image that remain detectable even when the image is rotated, scaled, or slightly distorted. For fingerprints, this is essential because no two scans of the same finger are perfectly aligned. The algorithm builds what is essentially a "feature map" of the fingerprint—a representation that captures its unique structure while ignoring variations in lighting, pressure, or finger placement. This is conceptually similar to how a CA kernel extracts meaningful patterns from a grid regardless of absolute positions.

```python
#Written with assistance from ChatGPT
def compare_fingerprints(path1, path2):
    """
    Takes in two fingerprint images and ensures they are in the correct format. Then uses SIFT to find
    unique points in the fingerprints and uses a brute force matcher to compare them. Good matches are
    filtered using Lowe's ratio test and the number of good matches is returned.
    """
    image1 = load_fingerprint(path1)
    image2 = load_fingerprint(path2)

    sift = cv2.SIFT_create()

    kp1, des1 = sift.detectAndCompute(image1, None)
    kp2, des2 = sift.detectAndCompute(image2, None)

    if des1 is None or des2 is None:
        return 0

    bf = cv2.BFMatcher()

    matches = bf.knnMatch(des1, des2, k=2)

    good_matches = []
    for m, n in matches:
        if m.distance < 0.75 * n.distance:
            good_matches.append(m)

    return len(good_matches)
```

**How Matches Are Scored:** After SIFT generates feature descriptors for both fingerprints, the system uses a brute-force matcher to find potential correspondences between them. However, not all matches are reliable. The Lowe's ratio test (the 0.75 factor in the code) acts as a quality filter: it compares the best candidate match against the second-best candidate. If the best match is significantly better than the second-best (by at least 25%), it is considered a "good match." This filtering step dramatically reduces false positives, ensuring that only distinctive feature pairs contribute to the final similarity score. The total count of good matches becomes the score—higher scores indicate more structural similarity between the two fingerprints.

```python
def find_matching_user(input_fingerprint_path, database, threshold=70):
    """
    Takes a fingerprint as input and compares it to all fingerprints in the database,then finds and returns the best match based on a threshold
    value.
    """
    user_found = None
    highest_score = -1

    for user in database:
        stored_path = user.get("fingerprint_path")

        score = compare_fingerprints(input_fingerprint_path, stored_path)

        if score > highest_score:
            highest_score = score
            user_found = user

    if highest_score >= threshold:
        return user_found, highest_score

    return None, highest_score
```

**Making the Decision:** The system compares the input fingerprint against every fingerprint in the database, recording the highest match score encountered. A threshold value (default 70) determines whether the best match is reliable enough for identification. This threshold acts much like a critical value in percolation theory: below it, connections are too weak to trust; above it, we have confidence that a genuine match exists. Setting the threshold appropriately involves a trade-off—lower values risk false matches (identifying the wrong person), while higher values risk false rejections (failing to identify a legitimate user).

```python
def show_match(input_path, matched_path):
    """
    Asked ChatGPT to write a visual demo function that displays the input fingerprint and the matched fingerprint side by side.
    """
    img1 = cv2.imread(input_path, cv2.IMREAD_GRAYSCALE)
    img2 = cv2.imread(matched_path, cv2.IMREAD_GRAYSCALE)

    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(img1, cmap='gray')
    plt.title("Input Fingerprint")
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(img2, cmap='gray')
    plt.title("Matched Fingerprint")
    plt.axis('off')

    plt.show()
```

```python
def find_match(path, dataset):
  """
  Takes a fingerprint as input and compares it to all fingerprints in the database,then determines whether there is a valid match. If valid match
  is found, then an image is displayed showing the input fingerprint and the matched fingerprint, and the matched user's full name and email
  is printed.
  """
  match, score = find_matching_user(path, dataset)

  if match:
      show_match(path, match["fingerprint_path"])
      print(f"Match found with user: {match['full_name']}")
      print(f"Email: {match['email']}")
  else:
      print("No match found")
```

```python
def plot_match_confidence(scores, dataset, threshold=70):
    """Displays a bar chart of match scores for all users in the database"""
    user_names = [user['full_name'] for user in dataset]

    plt.figure(figsize=(12, 6))
    bars = plt.bar(user_names, scores, color='blue')

    for i, score in enumerate(scores):
        if score >= threshold:
            bars[i].set_color('green')

    plt.axhline(y=threshold, color='red', linestyle='--', label=f'Threshold ({threshold})')
    plt.xlabel('Users')
    plt.ylabel('Match Score')
    plt.title('Fingerprint Match Scores Across Database')
    plt.xticks(rotation=45, ha='right')
    plt.legend()
    plt.tight_layout()
    plt.show()


def find_match_with_confidence(path, dataset):
    all_scores = []
    for user in dataset:
        stored_path = user.get("fingerprint_path")
        score = compare_fingerprints(path, stored_path)
        all_scores.append((user, score))

    all_scores.sort(key=lambda x: x[1], reverse=True)
    best_user, best_score = all_scores[0]

    scores_list = [score for _, score in all_scores]
    plot_match_confidence(scores_list, dataset, threshold=70) #Red line shows the threshold

    if best_score >= 70:
        print(f"Match found with user: {best_user['full_name']} (Score: {best_score})")
        print(f"Email: {best_user['email']}")
        print(f"Phone: {best_user['phone']}")
    else:
        print(f"No match found. Best score was {best_score}, below threshold 70.")

    return best_user if best_score >= 70 else None, best_score


    test_fp = dataset[3]["fingerprint_path"]
find_match_with_confidence(test_fp, dataset)
```
